# 1. 通过专业大模型厂商专用接口调用

调用DeepSeek模型

```python
import os
# 引入DeepSeeek模型api类
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv

# 需要显示加载配置文件
load_dotenv()  # 默认读取当前目录的 .env

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 初始化模型
llm_ds = ChatDeepSeek(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL
)

# 调用
res = llm_ds.invoke("你是谁？")
print(res)
```

# 4. 使用langchain1.x的统一方式

使用 init_chat_model()，官方文档：https://reference.langchain.com/python/langchain/chat_models/base/init_chat_model

```python

import os
from langchain.chat_models import init_chat_model

from dotenv import load_dotenv
# 需要显示加载配置文件
load_dotenv()  # 默认读取当前目录的 .env

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 初始化模型
llm_ds = init_chat_model(
    model="deepseek-flash",
    model_provider="deepseek",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL
)


res = llm_ds.invoke("1+1=?")
print(res)
```
